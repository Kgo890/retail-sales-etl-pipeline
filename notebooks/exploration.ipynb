```python
import numpy as np


# account adjustment
    # unitprice < 0 
# cancellation
    # InvoiceNo starts with "C"
# inventory adjustment 
    # quantity < 0 
    # unitprice == 0  
    # customerid is na
# zero-value transaction 
    # quantity > 0 
    # unitprice == 0

normalize_transaction_col = raw_transactions.copy()

conditions = [
    # account adjustment
    normalize_transaction_col['UnitPrice'] < 0,

    # cancellation
    normalize_transaction_col['InvoiceNo'].str.startswith('C', na=False),

    # inventory adjustment
    (normalize_transaction_col['Quantity'] < 0) & 
    (normalize_transaction_col['UnitPrice']  == 0) & 
    (normalize_transaction_col['CustomerID'].isna() == True),

    # zero-value transactions
    (normalize_transaction_col['Quantity'] > 0) & 
    (normalize_transaction_col['UnitPrice']  == 0)


]

conditions_categories = [
    'Account Adjustment',
    'Cancellation',
    'Inventory Adjustment',
    'Zero-Value Transaction'
]

normalize_transaction_col['TransactionType'] = np.select(conditions,conditions_categories, default='Sale')


print(normalize_transaction_col.head())

print(f"The number of rows that are account adjustment: {len(normalize_transaction_col[normalize_transaction_col['TransactionType'] == 'Account Adjustment'])}")
print(f"The number of rows that are cancellation: {len(normalize_transaction_col[normalize_transaction_col['TransactionType'] == 'Cancellation'])}")
print(f"The number of rows that are inventory adjustment: {len(normalize_transaction_col[normalize_transaction_col['TransactionType'] == 'Inventory Adjustment'])}")
print(f"The number of rows that are zero-value transaction: {len(normalize_transaction_col[normalize_transaction_col['TransactionType'] == 'Zero-Value Transaction'])}")
print(f"The number of rows that are sales: {len(normalize_transaction_col[normalize_transaction_col['TransactionType'] == 'Sale'])}")
```

```python
duplicated_normalize_transaction_col_with_keep = normalize_transaction_col[normalize_transaction_col.duplicated(keep=False)]
print(duplicated_normalize_transaction_col_with_keep.head(),"\n")
print(len(duplicated_normalize_transaction_col_with_keep),"\n")

duplicated_normalize_transaction_col = normalize_transaction_col[normalize_transaction_col.duplicated()]

print(duplicated_normalize_transaction_col.head(),"\n")
print(len(duplicated_normalize_transaction_col))
print(duplicated_normalize_transaction_col['TransactionType'].value_counts())

```

```python
duplicate_mask = normalize_transaction_col.duplicated(subset=['InvoiceNo','StockCode','Quantity'], keep = False)
all_duplicates = normalize_transaction_col[duplicate_mask]

intra_invoice_counts = (
    all_duplicates.groupby(['InvoiceNo','StockCode','Quantity']).size().reset_index(name = 'AppearanceCount')
                        )


same_invoice_clusters = intra_invoice_counts[intra_invoice_counts['AppearanceCount'] > 1]

cross_invoice_counts = (
    all_duplicates.groupby(['StockCode', 'Quantity'])['InvoiceNo']
    .nunique()
    .reset_index(name='UniqueInvoiceCount')
)

different_invoice_spans = cross_invoice_counts[cross_invoice_counts['UniqueInvoiceCount'] > 1]



print(f"Total rows involved in duplicates: {len(all_duplicates)}")
print(f"Distinct items repeated WITHIN the same invoice: {len(same_invoice_clusters)}")
print(f"Distinct items repeated ACROSS different invoices: {len(different_invoice_spans)}")

print("\n--- Example of Clustered Duplicates (Same Invoice) ---")
print(same_invoice_clusters.head())

print("\n--- Example of Spanned Duplicates (Different Invoices) ---")
print(different_invoice_spans.head())
```

```python
fact_sale = r"C:\Users\Kaden\retail_sales_etl_pipeline\data\processed\fact_sales.csv"
fact_adjustment = r"C:\Users\Kaden\retail_sales_etl_pipeline\data\processed\fact_adjustments.csv"
fact_cancellations = r"C:\Users\Kaden\retail_sales_etl_pipeline\data\processed\fact_cancellations.csv"
non_sale = r"C:\Users\Kaden\retail_sales_etl_pipeline\data\processed\non_sales.csv"

fact_sale_df = pd.read_csv(fact_sale)
fact_adjustment_df = pd.read_csv(fact_adjustment)
fact_cancellations_df = pd.read_csv(fact_cancellations)
non_sale_df = pd.read_csv(non_sale)

print(f"Total number of rows in fact_sale_df: {len(fact_sale_df)}")



def unique_combination_count(df):
    combination_count = (
        df[
            ['InvoiceNo', 'StockCode','Description','Quantity', 'InvoiceDate', 'UnitPrice']
        ]
        .drop_duplicates()
        .shape[0]
    )
    

    return combination_count

def unique_dupilcates(df):
    duplicates = ( 
        df[['InvoiceNo', 'StockCode', 'Description', 'Quantity', 'InvoiceDate', 'UnitPrice']]
        .duplicated(keep=False)
    )
    duplicate_rows = df[duplicates]
    return duplicate_rows

print(f"\nTotal number of the unique combinations for fact_sale {unique_combination_count(fact_sale_df)}")
print(f"\nTotal number of the unique combinations for fact_adjustment {unique_combination_count(fact_adjustment_df)}")
print(f"\nTotal number of the unique combinations for fact_cancellations {unique_combination_count(fact_cancellations_df)}")
print(f"\nTotal number of the unique combinations for non_sale {unique_combination_count(non_sale_df)}")



print(f"\nTotal number of duplicates of the combination for fact_sale: {len(unique_dupilcates(fact_sale_df))}")
print(f"Total number of duplicates of the combination for fact_adjustment: {len(unique_dupilcates(fact_adjustment_df))}")
print(f"Total number of duplicates of the combination for fact_cancellations: {len(unique_dupilcates(fact_cancellations_df))}")
print(f"Total number of duplicates of the combination for non_sale: {len(unique_dupilcates(non_sale_df))}")



```
